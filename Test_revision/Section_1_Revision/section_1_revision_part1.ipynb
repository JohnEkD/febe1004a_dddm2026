## **1. Python as a calculator**

An **expression** is anything that produces a value. Python evaluates the expression and, if it is
the last line in a cell, displays the result. Nothing is stored unless you give it a name.

| Operator | Meaning | Example | Value |
|---|---|---|---|
| `+` | addition | `2 + 9` | `11` |
| `-` | subtraction | `9 - 2` | `7` |
| `*` | multiplication | `2 * 4` | `8` |
| `/` | division | `2 / 9` | `0.2222...` |
| `**` | exponentiation | `2 ** 4` | `16` |
| `( )` | grouping | `(2 + 3) * 9` | `45` |

Order of operations: `**` first, then `*` and `/`, then `+` and `-`. Brackets override this.
Every operator needs a value on both sides, and `**` is two asterisks with no space between them.

````python
# Precedence: multiplication happens before addition
2 + 3 * 9
````

````python
# Brackets change the order
(2 + 3) * 9
````

**Watch out.** Only the *last* line of a cell is displayed. The cell below computes two values
but shows only the second one. Nothing is lost if you assign it a name, which is section 4.

````python
# Only the second value appears
2 + 3
2 * 3
````

## **2. `int` and `float`**

| Type | What it is | Written as | Produced by |
|---|---|---|---|
| `int` | whole number, exact, unlimited size | `2024` | `+` `-` `*` `**` on ints |
| `float` | decimal number, approximate, about 17 digits | `2024.0` | **any** use of `/` |

Division always produces a `float`, even when it divides evenly. A trailing `.0` in the output is
how you spot one. Because a `float` keeps only about 17 digits, arithmetic on floats is very
slightly wrong, so two floats that should be equal often are not.

````python
# Divides evenly, but the result is still a float
4048 / 2
````

````python
# Square root and squaring should cancel, but do not quite
(13 ** 0.5) ** 2
````

That last result is why you never test two floats for exact equality. Ints have no such
problem: they are exact and have no size limit, so `1234567 ** 20` is computed in full.

````python
# An int with 120-odd digits, exactly right
1234567 ** 20
````

## **3. Strings**

A string is a snippet of text in quotes. Single and double quotes behave identically, so use double
quotes when the text itself contains an apostrophe.

| Operation | Form | Example | Value |
|---|---|---|---|
| join | `string + string` | `'Ben' + ' 10'` | `'Ben 10'` |
| repeat | `string * int` | `'ha' * 3` | `'hahaha'` |

`+` needs a string on both sides. `'Ben' + 10` is a `TypeError`, because Python will not guess
whether you meant text or arithmetic. Convert the number first, which is section 6.

````python
# Double quotes, because the text has an apostrophe
"I don't ever want to miss class"
````

````python
# Joining needs the spacing to be part of one of the strings
"Ben" + " 10"
````

## **4. Names**

```
name = value
name = expression
```

Read `a = 2` aloud as "assign two to a", never as "a equals two". The right side is evaluated
first, and the name is then attached to the value that came out. Assignment displays nothing,
because it is an instruction rather than an expression.

The rule that catches everyone: **a name stores a value, not a formula.** Once `total = a + b` has
run, Python keeps the number and forgets where it came from. Changing `a` afterwards does not
change `total`. Only re-running the assignment does.

````python
# total is computed once, from the values of a and b at that moment
a = 2
b = 3
total = a + b
a = 10
total
````

`total` is still 5. If you expected 13, you were reading the assignment as a live formula.

A quoted word is a string. The same word unquoted is a name, and Python will look for an assignment
that does not exist and raise a `NameError`. That error almost always means a missing pair of quotes
or a cell you have not run yet.

````python
# The quoted version is text;
'FEBE1004A'

# The unquoted version would be a NameError
# FEBE1004A
````

Names also make a calculation readable. Compare `40 * 52 * 16.5` against the version below:
the numbers are the same, but only one of them says what it is computing.

````python
# Yearly earnings at a minimum wage of 16.50 an hour
hourly_minimum_wage = 16.50
hours_per_week = 40
weeks_per_year = 52
weekly_wages = hours_per_week * hourly_minimum_wage
weekly_wages * weeks_per_year
````

## **5. Functions and call expressions**

```
function(argument)
function(argument1, argument2)
function(argument, name=value)
```

The value inside the brackets is the **argument**. It may be an expression, in which case it is
evaluated first. Some functions accept a different number of arguments depending on what you want.

| Function | Form | Example | Value |
|---|---|---|---|
| `abs` | `abs(number)` | `abs(-5)` | `5` |
| `min` | `min(a, b)` | `min(14, 15)` | `14` |
| `round` | `round(number)` | `round(123.456)` | `123` |
| `round` | `round(number, ndigits)` | `round(123.456, 1)` | `123.5` |

`ndigits=1` is a **named argument**: you say which parameter you mean instead of relying on
position. To find out what the parameters are called, run `round?` in a cell, or `help(round)`
if the `?` panel does not display well.

````python
# The argument is evaluated before the function sees it
day_temp = 52
night_temp = 47
abs(night_temp - day_temp)
````

````python
# Positional and named arguments give the same result here
round(123.456, 1)
````

Python does not load everything by default. `import` makes a module available, and the dot means
"inside": `math.sqrt` is the `sqrt` function inside the `math` module. You will meet this constantly
from here on, in `np.average` and `np.arange`.

````python
# sqrt lives inside math, so math must be imported first
import math
math.sqrt(2)
````

## **6. Converting between types**

| Form | Converts to | Works on | Example |
|---|---|---|---|
| `int(x)` | `int` | floats, whole-number strings | `int('3')` gives `3` |
| `float(x)` | `float` | ints, numeric strings | `float('3.0')` gives `3.0` |
| `str(x)` | `str` | anything | `str(10)` gives `'10'` |

Two traps live here. First, `int` **truncates towards zero**, it does not round: `int(1.6)` is 1,
while `round(1.6)` is 2. They are different jobs and it is worth saying which one you want.
Second, `int('3.14')` fails, because `int` cannot parse a string containing a decimal point.
Go via float instead: `int(float('3.14'))`.

````python
# Truncation against rounding, on the same number
int(1.6), round(1.6)
````

````python
# A numeric string with a decimal point needs float first
int(float('7.99'))
````

## **7. Every value has a type**

```
type(value)
```

`type` returns the kind of value you are holding, not the value itself. It works on anything,
including names and functions, and it is the first thing to reach for when something behaves
unexpectedly. Most Section 1 errors are a type in a place you did not expect.

````python
# Multiplication keeps ints; division does not
type(1012 * 2), type(4048 / 2)
````

````python
# Converting a number to a string lets it join to other text
type('Ben ' + str(10))
````

## **9. Self-check**

Answer these without scrolling back. Reveal each answer only after you have committed to one.

**Q1.** What does `2 + 3 * 9 ** 2` evaluate to?

<details>
<summary><strong>Answer</strong></summary>

<code>245</code>. Exponentiation runs first, giving 81, then multiplication gives 243, then addition gives 245. If you got 2025 you applied the operators strictly left to right.

</details>

**Q2.** True or false: a cell whose last line is a bare name, such as `yearly_wages`, displays the stored value when it runs.

<details>
<summary><strong>Answer</strong></summary>

<strong>True.</strong> Python displays the value of the last expression in a cell when that expression is not an assignment. A bare name is an expression, so its value is shown.

</details>

**Q3.** After running `x = 5`, then `y = x * 2`, then `x = 100`, what is the value of `y`?

<details>
<summary><strong>Answer</strong></summary>

<code>y</code> is <strong>10</strong>. When <code>y = x * 2</code> ran, <code>x</code> was 5, so 10 was stored. A name holds a value, not a formula, so reassigning <code>x</code> afterwards changes nothing.

</details>

**Q4.** What does `round(2.567, 2)` return, and what is the `2` called?

<details>
<summary><strong>Answer</strong></summary>

<code>2.57</code>. The second argument is the number of decimal places, and its parameter name is <code>ndigits</code>, so <code>round(2.567, ndigits=2)</code> is the same call written with a named argument.

</details>

**Q5.** What does `int(float('7.99'))` return? What does `int('7.99')` return?

<details>
<summary><strong>Answer</strong></summary>

<code>int(float('7.99'))</code> returns <strong>7</strong>: the string becomes the float 7.99, which <code>int</code> then truncates towards zero. <code>int('7.99')</code> raises a <code>ValueError</code>, because <code>int</code> cannot parse a string containing a decimal point.

</details>

**Q6.** Why does `type(4048 / 2)` report `float` when the division comes out even?

<details>
<summary><strong>Answer</strong></summary>

Any use of <code>/</code> produces a float, without exception. Python does not inspect the result to decide. The trailing <code>.0</code> in the output is the visible sign of it.

</details>

**Q7.** `'ha' * 3` gives `'hahaha'`. What does `'ha' + 3` give?

<details>
<summary><strong>Answer</strong></summary>

A <code>TypeError</code>. <code>*</code> between a string and an int repeats the string, but <code>+</code> requires the same type on both sides. Convert first: <code>'ha' + str(3)</code> gives <code>'ha3'</code>.

</details>

**Q8.** You run a cell and get `NameError: name 'wages' is not defined`, but you are sure you wrote that assignment. What are the two likely causes?

<details>
<summary><strong>Answer</strong></summary>

Either the cell containing the assignment has not been run in this session, or the name is spelled differently in the two places. Running the notebook top to bottom rules out the first, which is why <em>Run All Above</em> is the usual first move.

</details>
