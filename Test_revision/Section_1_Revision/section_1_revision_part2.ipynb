## **1. An array holds many values under one name**

```
make_array(value1, value2, value3, ...)
```

An array keeps its values in the order you gave them

**Example:** Create an array, `my_array`, with four values, their positions (indexes) are 0, 1, 2 and 3

````python

my_array = make_array(1, 2, 3, 4)
my_array
````

---

```
array.item(index)
```
and `.item(index)` pulls one out by position.
Positions start at **zero**, so the first element is `item(0)`. An array of length *n* has valid
positions `0` to `n - 1`, and asking for `n` raises an `IndexError`.

Counting from zero is the single most common source of off-by-one mistakes in this course. The
useful phrasing is that the index is not "which one" but "how far along from the start".

- Four values, so the valid positions are 0, 1, 2 and 3

````python
my_array.item(0)
````

- The last element is at index 3, not 4

````python
my_array.item(3)
````

## **2. Arithmetic happens element by element**

| Form | Meaning | Example | Value |
|---|---|---|---|
| `array + number` | apply to every element | `make_array(1,2,3) + 2` | `array([3, 4, 5])` |
| `array * number` | apply to every element | `make_array(1,2,3) * 2` | `array([2, 4, 6])` |
| `array + array` | pair up position by position | `make_array(1,2) + make_array(3,4)` | `array([4, 6])` |

Arithmetic on an array applies to every element at once, so you never write a loop. The result is a
**new** array, and the original is unchanged unless you assign the result to a name.

Note that `array * 2` **scales**. It does not repeat. Repetition is what `*` does to a string, and
the same symbol behaving differently on two types is worth holding on to.

- The example below is scaling, not repeating

````python
make_array(1, 2, 3) * 2
````

- Two arrays are paired position by position

````python
make_array(1, 2, 3, 4) + make_array(4, 5, 6, 7)
````

**Watch out.** Pairing by position only works if both arrays are the same length. A mismatch
raises a `ValueError` about shapes, and every arithmetic operator has the same requirement.
The cell below is the error you will see, left in deliberately.

- Four elements against three: this raises a ValueError

````python
my_array + make_array(8, 9, 10)
````

## **3. Functions that collapse an array to one value**

| Form | Returns | Example | Value |
|---|---|---|---|
| `sum(array)` | one number | `sum(make_array(1,2,3,4))` | `10` |
| `np.average(array)` | one number | `np.average(make_array(1,2,3,4))` | `2.5` |
| `len(array)` | one number | `len(make_array(1,2,3,4))` | `4` |

These are a different shape of function from `array + 2`. That one gives back an array of the same
length; these give back a single number. Keeping the two shapes apart tells you what you are
holding after each step, which matters as soon as you start chaining operations in the tables
notebook.

`np.average` lives inside numpy, hence the `np.`. `sum` and `len` are built into Python.

````python
# Array in, single number out
my_array = make_array(1, 2, 3, 4)
sum(my_array), np.average(my_array), len(my_array)
````

The average is the sum divided by the length, so the two lines below agree. Writing it out
occasionally is a good way to check that a function is doing what you assume.

````python
# The definition of an average, spelled out
sum(my_array) / len(my_array)
````

## **4. Ranges are sequences of consecutive numbers**

```
np.arange(stop)
np.arange(start, stop)
np.arange(start, stop, step)
```

| Call | Value |
|---|---|
| `np.arange(7)` | `array([0, 1, 2, 3, 4, 5, 6])` |
| `np.arange(5, 11)` | `array([5, 6, 7, 8, 9, 10])` |
| `np.arange(0, 7, 3)` | `array([0, 3, 6])` |
| `np.arange(20, 0, -2)` | `array([20, 18, ..., 2])` |

`stop` is **excluded**, so `np.arange(7)` stops at 6 and gives seven elements starting from zero.
`start` defaults to 0 and `step` defaults to 1. The step may be a decimal or negative.

The exclusion of `stop` pairs neatly with counting from zero: `np.arange(n)` gives you exactly the
valid positions of an array of length *n*.

````python
# Seven elements, ending at 6 because stop is excluded
np.arange(7)
````

````python
# A negative step counts downwards, still excluding stop
np.arange(20, 0, -2)
````

Named arguments work here too, and they make a three-argument call much easier to read back.
Use `np.arange?` if you forget the order.

````python
# The same call, written so the reader can see which is which
np.arange(start=0, stop=7, step=3)
````

## **6. Self-check**

Answer these without scrolling back. Reveal each answer only after you have committed to one.

**Q1.** You have `a = make_array(10, 20, 30, 40, 50)`. What does `a.item(4)` return? What does `a.item(5)` return?

<details>
<summary><strong>Answer</strong></summary>

<code>a.item(4)</code> returns <strong>50</strong>, the last element, because positions start at zero. <code>a.item(5)</code> raises an <code>IndexError</code>: an array of length 5 has positions 0 to 4 only.

</details>

**Q2.** What does `make_array(3, 6, 9) / 3` evaluate to?

<details>
<summary><strong>Answer</strong></summary>

<code>array([1., 2., 3.])</code>. Division is applied to every element. The results are floats because <code>/</code> always produces a float, which is why the output shows <code>1.</code> rather than <code>1</code>.

</details>

**Q3.** How many elements does `np.arange(2, 14, 4)` contain, and what are they?

<details>
<summary><strong>Answer</strong></summary>

Three: <code>[2, 6, 10]</code>. Starting at 2 and stepping by 4 gives 2, 6, 10, and the next value would be 14, which is the excluded stop.

</details>

**Q4.** What does `make_array(1, 2, 3) + np.arange(start=1, stop=4)` give?

<details>
<summary><strong>Answer</strong></summary>

<code>array([2, 4, 6])</code>. <code>np.arange(start=1, stop=4)</code> is <code>[1, 2, 3]</code> because stop is excluded, and adding it element-wise to <code>[1, 2, 3]</code> doubles each value. If you expected a length error, you read arange as going up to and including 4.

</details>

**Q5.** `'ha' * 3` repeats the text. What does `make_array(1, 2, 3) * 3` do, and why is that not inconsistent?

<details>
<summary><strong>Answer</strong></summary>

It scales, giving <code>array([3, 6, 9])</code>. The behaviour of an operator depends on the type of the values it sits between, so <code>*</code> means repetition for a string and element-wise multiplication for an array. When an operator surprises you, check the types first with <code>type</code>.

</details>

**Q6.** Why does `make_array(1, 2, 3, 4) * make_array(2, 3)` fail, and what would you check first?

<details>
<summary><strong>Answer</strong></summary>

Element-wise arithmetic pairs values by position, so both arrays must have the same length. Four against two cannot pair, and numpy raises a <code>ValueError</code> about shapes. Check <code>len</code> on both arrays before combining them.

</details>

**Q7.** `sum(my_array)` and `my_array + my_array` both involve addition. What is different about what each gives back?

<details>
<summary><strong>Answer</strong></summary>

<code>sum</code> collapses the array to a <strong>single number</strong>, adding across positions. <code>my_array + my_array</code> returns an <strong>array of the same length</strong>, adding at each position separately. Knowing which shape you are holding decides what you can legally do next.

</details>

## **7. Cheatsheet**

| Form | What it does | Returns |
|---|---|---|
| `make_array(1, 2, 3)` | builds an array from values | array |
| `array.item(0)` | the value at a position, counting from zero | one value |
| `array + 2` | applies to every element | array |
| `array * 2` | scales every element | array |
| `array + array` | pairs by position; lengths must match | array |
| `sum(array)` | adds every element | one number |
| `np.average(array)` | the mean | one number |
| `len(array)` | how many elements | one number |
| `np.arange(7)` | 0 up to but excluding 7 | array |
| `np.arange(5, 11)` | 5 up to but excluding 11 | array |
| `np.arange(20, 0, -2)` | counts down, excluding 0 | array |
| `np.arange?` | shows the documentation | a panel |
