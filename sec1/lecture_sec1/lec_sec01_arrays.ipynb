## 1. Arrays are sequences of values

**Syntax - arrays**

```
make_array(value1, value2, value3, ...)
array.item(index)
```

- An array holds many values under one name, in the order you gave them.
- `.item(index)` pulls out one value by position.
- Positions start at **zero**. The first element is `item(0)`, not `item(1)`.
- An array of length *n* has valid positions `0` to `n - 1`. Asking for `n` is an `IndexError`.

````python
my_array = make_array(1, 2, 3, 4)
````

````python
my_array
````

````python
type(my_array)
````

````python
my_array.item(1)
````

````python
my_array.item(0)
````

An array of length 4 has positions `0`, `1`, `2`, `3`. The line below causes an *index error* because there is no position `4`.

````python
my_array.item(4)
````

## 2. Array arithmetic happens element by element

**Syntax - array arithmetic**

| Form | Meaning | Example | Value |
|---|---|---|---|
| `array + number` | apply to every element | `make_array(1,2,3) + 2` | `array([3, 4, 5])` |
| `array * number` | apply to every element | `make_array(1,2,3) * 2` | `array([2, 4, 6])` |
| `array + array` | pair up position by position | `make_array(1,2) + make_array(3,4)` | `array([4, 6])` |

- Arithmetic on an array applies to **every element**. You never write a loop.
- Two arrays must be the **same length**, or you get a shape error.
- These produce a **new** array. The original is unchanged unless you assign the result.
- Note `array * 2` **scales**. It does not repeat - that is what `*` does to a string.

````python
my_array + 2
````

````python
my_array * 2
````

````python
my_array ** 2
````

````python
my_array_2 = make_array(4,5,6,7)
````

````python
my_array + my_array_2
````

---
Element-wise arithmetic pairs values by position, so both arrays must be the same length. The line below causes a *value error* because `my_array` has 4 elements and `make_array(8, 9, 10)` has 3.

````python
my_array + make_array(8,9,10)
````

---
The same mismatch with `*` instead of `+`. Every arithmetic operator has the same length requirement.

````python
my_array * make_array(8,9,10)
````

## 3. Functions on arrays

**Syntax - functions that collapse an array**

| Form | Returns | Example | Value |
|---|---|---|---|
| `sum(array)` | one number | `sum(make_array(1,2,3,4))` | `10` |
| `np.average(array)` | one number | `np.average(make_array(1,2,3,4))` | `2.5` |
| `len(array)` | one number | `len(make_array(1,2,3,4))` | `4` |

- These take an array and give back a **single value** - a different shape of
  function from `array + 2`, which gives back an array.
- `np.average` lives inside numpy, hence the `np.`. `sum` and `len` are built into Python.

````python
my_array = make_array(1, 2, 3, 4)
````

````python
sum(my_array)
````

````python
np.average(my_array)
````

````python
len(my_array)
````

## 4. Ranges are sequences of consecutive numbers

**Syntax - ranges**

```
np.arange(stop)
np.arange(start, stop)
np.arange(start, stop, step)
```

| Call | Value |
|---|---|
| `np.arange(7)` | `array([0, 1, 2, 3, 4, 5, 6])` |
| `np.arange(5, 11)` | `array([5, 6, 7, 8, 9, 10])` |
| `np.arange(0, 7, 3)` | `array([0, 3, 6])` |
| `np.arange(20, 0, -2)` | `array([20, 18, ..., 2])` |

- `stop` is **excluded**. `np.arange(7)` stops at 6.
- `start` defaults to `0`, `step` defaults to `1`.
- `step` may be a decimal or negative.

````python
make_array(0,1,2,3,4,5,6)
````

````python
np.arange(7)
````

````python
np.arange(stop=7)
````

````python
np.arange?
````

````python
np.arange(start=0, stop=7, step=3)
````

````python
np.arange(5, 11)
````

````python
np.arange(0, 1, 0.1)
````

````python
np.arange(20,0,-2)
````
