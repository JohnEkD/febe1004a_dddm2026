## 1. Python as a calculator

**Syntax - arithmetic operators**

| Operator | Meaning | Example | Value |
|---|---|---|---|
| `+` | addition | `2 + 9` | `11` |
| `-` | subtraction | `9 - 2` | `7` |
| `*` | multiplication | `2 * 4` | `8` |
| `/` | division | `2 / 9` | `0.2222...` |
| `**` | exponentiation | `2 ** 4` | `16` |
| `( )` | grouping | `(2 + 3) * 9` | `45` |

- Every operator needs a value on **both** sides. `* 4` on its own is not valid.
- `**` is two asterisks with **no space** between them.
- Order of operations: `**` first, then `*` and `/`, then `+` and `-`. Brackets override this.

An **expression** is anything that evaluates to a value. `2`, `2 + 9`, and `(2 + 3) * 9`
are all expressions. Every line in this section is an expression, and Python displays
the value of the last expression in a cell.

Python as a calculator:

## 4. Names

**Syntax - assignment**

```
name = value
name = expression
```

- The left side is always a **name**, never an expression.
- The right side is evaluated **first**; the name is then attached to the resulting value.
- Assignment produces no output. It is an instruction, not an expression.

Read `a = 2` aloud as "**assign** two **to** a", or "a **gets** two" - never "a equals two".

````python
a = 2
````

````python
a
````

## 5. Functions, call expressions and arguments

**Syntax - call expressions**

```
function(argument)
function(argument1, argument2)
function(argument, name=value)
```

- The value inside the brackets is the **argument**. It may itself be an expression,
  in which case it is evaluated **first**.
- `name=value` is a **named argument**: you say which parameter you mean rather than
  relying on position.
- `function?` displays the documentation, including the parameter names.

| Function | Form | Example | Value |
|---|---|---|---|
| `abs` | `abs(number)` | `abs(-5)` | `5` |
| `min` | `min(a, b)` | `min(14, 15)` | `14` |
| `round` | `round(number)` | `round(123.456)` | `123` |
| `round` | `round(number, ndigits)` | `round(123.456, 1)` | `123.5` |

````python
abs(-5)
````

````python
abs(1 - 3)
````

Names can help us with inputs (called **arguments** in computer science jargon) to functions!

````python
day_temp = 52
night_temp = 47
abs(night_temp - day_temp)
````

Functions can take multiple arguments.

````python
min(14, 15)
````

Some functions can be called with a single argument *or* with multiple arguments!

````python
round(123.456)
````

````python
round(123.456, 1)
````

````python
round(123.456, ndigits=1)
````

In the third line, `ndigits` is a **named argument**.

How did I know that `ndigits` would work as a name? You can use `?` to figure out everything you want to know about a function.

````python
round?
````

````python
round(number=123.456)
````

## 6. Converting between types

**Syntax - type conversion**

| Form | Converts to | Works on | Example |
|---|---|---|---|
| `int(x)` | `int` | floats, whole-number strings | `int('3')` ---> `3` |
| `float(x)` | `float` | ints, numeric strings | `float('3.0')` ---> `3.0` |
| `str(x)` | `str` | anything | `str(10)` ---> `'10'` |

- `int()` **truncates** toward zero. It does not round. `int(1.6)` is `1`.
- `round()` rounds. `round(1.6)` is `2`. These are different jobs.
- `int('3.14')` fails - go via float: `int(float('3.14'))`.

````python
int('3')
````

````python
int(float('3.0'))
````

````python
float(int('3'))
````

````python
str(10)
````

````python
'Ben ' + str(10)
````

````python
2 + float(8)
````
